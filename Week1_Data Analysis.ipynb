## **List Operations and Manipulations (mutable)**

* **Adding element in list:**


```python
list1=['Mango','apple','banana','1','2']
list1[1]='cherry'
print(list1)
```

```python
#append() function
my_list = [1, 2, 3, 4, 5, 6]
print(my_list)
my_list.append(7) #add as a single element
print(my_list)
my_list
my_list.append(['red','blue','green']) #add multiple elements
my_list
```

```python
#extend() function
my_list = [1, 2, 3, 4, 5, 6]
list1= [7,8,9]
my_list.extend(list1) #add as different elements
print(my_list)
```

```python
#insert() function
my_list = ['this', 'is', 'Module', 2024, 'Semester', 1]
print(my_list)
my_list.insert(2,'CMP7005') #add element at 2 index
print(my_list)
```

* **Deleting Element:**

```python
#del keyword
my_list = ['this', 'is', 'Module', 2024, 'Semester', 1]
print(my_list)
del my_list[5] #delete element at index 4
print(my_list)
```

```python
# pop() function
my_list = ['this', 'is', 'Module', 2024, 'Semester', 1]
print(my_list)
a = my_list.pop(1) #pop element from list
print('Popped Element:', a,)
print('List remaining: ', my_list)
```

```python
# remove() function
my_list = ['this', 'is', 'Module', 2024, 'Semester', 1]
print(my_list)
my_list.remove('Module') #remove element with value
print(my_list)
```

```python
# clear() function
my_list = ['this', 'is', 'Module', 2024, 'Semester', 1]
print(my_list)
my_list.clear() #empty the list
print(my_list)
```

* **Some more list operations:**

```python
mylist = [8, 20, 7, 20, 30]
# len() function
print("The lenght is:", len(mylist))
# min() function
print("The minimum is:",min(mylist))
# max() function
print("The maximum is:",max(mylist))
# index() method
print("The index of 20 is:", mylist.index(20))

mylist.sort()
print(mylist)
```
